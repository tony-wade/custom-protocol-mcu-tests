```python
CLK_PERIOD = 41
TIMER_CLOCK = 48000000
FIXED_POINT_MULTIPLIER = 1000
NS_PER_TICK_FIXED = (1000000000 * FIXED_POINT_MULTIPLIER) // TIMER_CLOCK

NORMAL_PERIOD = 750
NORMAL_HIGH = 284
START_STOP_PERIOD = 1400
START_STOP_HIGH = 1000

start_stop_positions = {6, 41}  # 定義 Start-Stop 位置
```

```python
CLK_reload = []
clk_state = 1  # 初始狀態

# CLK Schedule
for i in range(CLK_PERIOD * 2):
    if i in start_stop_positions:
        period, duration = START_STOP_PERIOD, START_STOP_HIGH
    elif (i + 1) in start_stop_positions:
        period, duration = NORMAL_PERIOD + 150, NORMAL_HIGH
    else:
        period, duration = NORMAL_PERIOD, NORMAL_HIGH

    if clk_state == 0:  # Low phase
        reload_value = 65536 - ((period - duration) * FIXED_POINT_MULTIPLIER + (NS_PER_TICK_FIXED >> 1)) // NS_PER_TICK_FIXED
    else:  # High phase
        reload_value = 65536 - (duration * FIXED_POINT_MULTIPLIER + (NS_PER_TICK_FIXED >> 1)) // NS_PER_TICK_FIXED

    CLK_reload.append(reload_value)

    clk_state ^= 1

# 轉換成 C 陣列格式
c_array = ", ".join(map(str, CLK_reload))
print(f"code uint16_t CLK_reload[{len(CLK_reload)}] = {{ {c_array} }};")
```
